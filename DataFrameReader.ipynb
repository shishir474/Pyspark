```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.master("local[*]").appName("ReadingAndWritingDataUsingSparkDF").getOrCreate()
```

```python
from pyspark.sql.types import *

data = [(1, "Shishir"), (2, "Rahul"), (3, "Ajay")]

schema = StructType([StructField("id", StringType()), StructField("name", StringType())])

# option 1: provided just column name. datatype of each column will be inferred by spark by scanning the data
df1 = spark.createDataFrame(data, schema = ["id", "name"])

# option 2: provided the schema explicitly
df2 = spark.createDataFrame(data,schema)
df1.show()
df1.printSchema()
```

```python
# Write the csv file at the mentioned location
# 3 imp parameters: saving mode, file format and location where you need to save
df2.write.mode("overwrite").option("header", True).format("csv").save("/Users/shishir/Pyspark/SampleData/df2.csv")
```

```python
# read the same csv file back into pyspark df
schema = StructType([StructField("id", IntegerType()), StructField("name", StringType())])

# Option 1: spark.read.csv("filepath")
csv_df = spark.read.option("header",True).csv("/Users/shishir/Pyspark/SampleData/df2.csv", schema)

# Other option using spark.read.format("csv").load(filepath)
csv_df2 = spark.read.schema(schema).option("header",True).format("csv").load("/Users/shishir/Pyspark/SampleData/df2.csv")
csv_df.show()
csv_df.printSchema()
```
