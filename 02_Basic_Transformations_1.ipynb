```python
# Small example for schema

schema_string = "name string, age int"

# in order to write this in native spark format - we need to import StructType, StructField, StringType, IntegerType
# from pyspark.sql.types module

from pyspark.sql.types import StructType, StructField, StringType, IntegerType

schema_spark = StructType([
        StructField("name",StringType(),False),
        StructField("age",IntegerType(),False)]
    )
# This is how you define schema in spark. However you can also use schema strings to do the same. Spark uses a method to implicitly convert the schema string into the spark's native schema.

# Why do we need to pass StringType() with () ?
# Bcoz spark expects an instance of the datatype class, not the class itself. StringType is a class that defines the string data type in Spark, and by calling StringType(), we create an instance of that class which can be used to specify the data type of a column in a DataFrame schema. If we were to pass StringType without the parentheses, we would be passing the class itself rather than an instance of the class, which would lead to an error when Spark tries to interpret it as a data type.
# Why does spark expect an instance of the datatype class instead of the class itself?
# Internally spark, stores the metadata about the type, serializes the schema, compares types across schema. All of this requires a fully constructed object and not just a reference to the class.
```

```python
# Columns and expressions

# We can select, manipulate or remove columns from dataframes and these operations are called expressions.
# Now to put expressions on columns, we need to have real values and for that we need rows, and to have that we need dataframes.
# That is why we cannnot manipulate individual columns out of dataframes. To manipulate a column we need a dataframe.

# There are many ways in which we can define or refer a column. To call a columnn explicilty, we can use col() or expr() from pyspark.sql.functions module 

from pyspark.sql.functions import col, expr

col("name")
```

```python
# This gives us the same col name reference, because any manipulation done on a column is considered an expression

expr("name")
```

```python
# to call a column from a dataframe, we can use . 

emp.salary
```

```python
# Another way to call salary column -- using square brackets

emp["salary"]
```

```python
# Select columns

# To make things easier, lets correlate our dataframe queries with sql queries
# select employee_id, name, age, salary from emp

emp_filtered = emp.select(col("employee_id"), emp.name, expr("age"), emp["salary"])

# nothing happens on spark UI - because select is a transformation
```

```python
# show dataframe (action)

# This is an action and this will trigger a job in the background

emp_filtered.show()
```

```python
# rename employee_id as emp_id & cast age as int
# select employee_id as emp_id, name, cast(age as int) as age, salary from emp_filtered

emp_casted = emp_filtered.select(expr("employee_id as emp_id"), col("name"), expr("cast(age as int) as age"), col("salary"))

# nothing happens as select is a transformation
# It is pretty evident that we can use expr() to write column expressions.
# Spark provides one better way - combination of select and expr: selectexpr()
```

```python
# show dataframe (action)

emp_casted.show()
```

```python
# Just to verify, if the columns have been casted
emp_casted.printSchema()
```

```python
emp_casted_1 = emp_filtered.selectExpr("employee_id as emp_id", "name", "cast(age as int) as age", "salary")

# advantage of selectExpr() is you dont need to call expr() explicitly for each expressions, because selectExpr() does the job
# you can directly write column names inside selectExpr()

# selectExpr() = select() + expr()
```

```python
emp_casted_1.show()
```

```python
emp_final = emp_casted.select(col("emp_id"), col("name"), col("age"), col("salary")).where("age > 30")
```

```python
# SHOW Dataframe(action)
emp_final.show()
```

```python
# Write the data back as CSV (action)

emp_final.write.format("csv").save("pyspark/output/2/emp.csv")
```
