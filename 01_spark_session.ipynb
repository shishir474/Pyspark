```python
# Spark Session
from pyspark.sql import SparkSession

# SparkSession is the driver process which is the entrypoint of our spark application
# spark is the SparkSession object 
spark = SparkSession.builder.appName('Spark Introduction').master("local[*]").getOrCreate()
```

```python
# create emp dataframe
emp = spark.createDataFrame(data = emp_data, schema = emp_schema)
```

```python
emp.printSchema()
```

```python
# Check no of partitions

# Spark reads data in form of partitions, so whatever dataframe we create would be built up in number of partitions
emp.rdd.getNumPartitions()
```

```python
# Show data (Action

# Actions are responsible for creating jobs -- check spark UI

emp.show()
```

```python
# Write our first transformation (emp salary > 50000) 

# dataframes are immutable -- we cannot change anything about the existing dataframes, but can create new dataframes out of it

emp_final = emp.where("salary > 50000")

# Nothing happens on the Spark UI because spark prefers lazy evauluation. So until an action is called like show() etc, no computations will be performed 
```

```python
# validate emp_final dataframe partitions

emp_final.rdd.getNumPartitions()
```

```python
# Write data as csv output (Action)

emp_final.write.format("csv").save("pyspark/output/emp.csv")
```
